```python
import gamdpy as gp
import h5py

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os.path

import matplotlib.pyplot as plt
from matplotlib.ticker import AutoMinorLocator, LogLocator
plt.style.use('physrev')

gp.select_gpu()
```

```python
NaCl = {
    "fractional_coordinates": [
        # Na
        [0.0, 0.0, 0.0],
        [0.5, 0.5, 0.0],
        [0.5, 0.0, 0.5],
        [0.0, 0.5, 0.5],

        # Cl
        [0.5, 0.0, 0.0],
        [0.0, 0.5, 0.0],
        [0.0, 0.0, 0.5],
        [0.5, 0.5, 0.5],
    ],
    "lattice_constants": [1.0, 1.0, 1.0]
}
conf = gp.Configuration(D=3)
conf.make_lattice(NaCl, cells=[6, 6, 6])

conf["q"] = np.tile([1, 1, 1, 1, -1, -1, -1, -1], 6 * 6 * 6)
```

```python
damping = 1
rcq = 3
cut_q = np.full((2,2), rcq)
n = 20
```

```python
electro_pot = gp.Electrostatics(damping=damping, cutoff=cut_q)
electro_pot.set_ewald(ncut=n)
```

```python
evaluator = gp.Evaluator(configuration, [electro_pot])
```

```python
evaluator.evaluate(configuration)
```

```python
charges, _ = configuration.get_charged_particles()
self_term = np.sum(charges**2) * damping / np.sqrt(np.pi)
```

```python
u = evaluator.configuration['U'] - self_term
```
